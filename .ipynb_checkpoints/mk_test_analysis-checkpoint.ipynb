## Permutations

```python
def perm_shared_genes(numGenes, sigList, numPerm):
    """
    numGenes: total number of genes (universe size)
    sigList:  list/tuple of sizes of each gene set (e.g. [100, 200, 300])
    numPerm:  number of permutations
    """
    permShared = []
    for _ in range(numPerm):
        perms = [set(np.random.choice(range(1, numGenes+1), size=s, replace=False)) for s in sigList]
        shared = set.intersection(*perms)
        permShared.append(len(shared))
    return permShared
```

```python
import math

## BP
# set_list = [set1, set2, set3, set4, set5]

## controls
set_list = [set6, set7, set8]

numGenes = 16_000
numPerm = 10_000

# 10+10+5+1=26

num_plots = sum([math.comb(len(set_list), k) for k in range(2, len(set_list)+1)])

plots_per_row = 5

nrows = math.ceil(num_plots/plots_per_row)
fig, axes = plt.subplots(nrows, plots_per_row, figsize=(plots_per_row*4, nrows*2.7))

axes = axes.flatten()

i = 0
for k in range(2, len(set_list)+1):  # k=2,3,4,5
    for combo in itertools.combinations(set_list, k):

        set_sizes = [len(s) for s in combo]
        observed_shared = len(set.intersection(*combo))
        permShared = perm_shared_genes(numGenes, set_sizes, numPerm)
        
        axi = axes[i]
        bins = np.arange(min(permShared), max(permShared)+2)
        sns.histplot(ax=axi, data=permShared, bins=bins, kde=False)

        axi.axvline(observed_shared, color='red', linestyle='-')
        axi.set_xlim(-0.1, 11)
        axi.set_ylim(0, 10000)
        
        label_txt = ', '.join(f'{l}:{n}' for l, n in zip('ABCDE', set_sizes))
        title = f"{k}-way: {label_txt}\nShared:{observed_shared}"
        axi.set_title(title, fontsize=10)
        axi.tick_params(axis='both', which='both', labelleft=True, labelbottom=True)
        i += 1
        print(i)

# Remove unused axes
for j in range(i, len(axes)):
    fig.delaxes(axes[j])

sns.despine(fig=None, ax=None, left=False, bottom=False, offset=None, trim=False)
fig.tight_layout()

plt.show()
# fig.savefig('permutations_bp_all_2to5way.v2.pdf', bbox_inches='tight')
fig.savefig('permutations_controls_all_2to3way.v2.pdf', bbox_inches='tight')
```

```python
np.random.seed(123)
gene_pool = np.arange(16_000)
num_iter = 10_000

fig, axes = plt.subplots(1, 3, figsize=(9, 2.3)) 
finches = [set1, set2, set3]

for i in range(len(finches)):
    s = finches[i]
    axi = axes[i]
    observed = len((s & set4) | (s & set5) | (set4 & set5))
    
    shared_counts = []
    for _ in range(num_iter):
        sa = np.random.choice(gene_pool, size=len(s), replace=False)
        sb = np.random.choice(gene_pool, size=len(set4), replace=False)
        sc = np.random.choice(gene_pool, size=len(set5), replace=False)
        combined = np.concatenate([sa, sb, sc])
        (unique, counts) = np.unique(combined, return_counts=True)
        n_shared = np.sum(counts >= 2)
        shared_counts.append(n_shared)

    bins = np.arange(min(shared_counts), max(shared_counts)+2)
    sns.histplot(ax=axi, data=shared_counts, bins=bins, kde=False)
    axi.axvline(observed, color='red', linestyle='-')
    axi.set_xlim(-1, 11)
    axi.set_ylim(0, 10000)
    
    p_value = np.mean(np.array(shared_counts) >= observed)
    axi.set_title(f"p-value:{p_value:.5f}")
        
sns.despine(fig=None, ax=None, left=False, bottom=False, offset=None, trim=False)
fig.tight_layout()

plt.show()
fig.savefig('permutations_bp_at_least_2.MKT.pdf', bbox_inches='tight')
```

```python
## Get genes and their AF by category
# requires all_DBs!
dbs = ['vidMac', 'vidCha', 'indInd']

## store genes by GO category in a dictionary
goterms_dict = {}
for db in dbs:
    af_files = glob(dir_path + 'MK_test_{}_ncbi/*genes.{}.lst'.format(db, db))
    for af_file in af_files:
        i = af_file.split('/')[-1].split('.')[0]
        with open(af_file, 'r') as inf:
            genes = inf.read().rstrip().split('\n')
            goterms_dict[(db, i)] = genes

## Make nonsyn AF dataframe
all_AFs = {'db': [],
           'genes': [],
           'AF': []
          }

for db_term in goterms_dict:
    db = db_term[0]
    term = db_term[1]
    
    # get corresponding nonsynonymous AFs from all_DFs
    dfi = all_DFs[db]
    af_i = ''.join(list(dfi[dfi['gene'].isin(goterms_dict[(db, term)])]['af_syn'])).split(',')
    all_genes_af_i = ''.join(list(dfi['af_syn'])).split(',')
    
    # drop empty elements
    af_i = ' '.join(af_i).split()
    all_genes_af_i = ' '.join(all_genes_af_i).split()

    # First, add AFs of all genes if they are not there yet
    if db not in all_AFs['db']:
        all_AFs['db'].extend([db] * len(all_genes_af_i))
        all_AFs['genes'].extend(['all_genes'] * len(all_genes_af_i))
        all_AFs['AF'].extend([float(af) for af in all_genes_af_i])
    
    # Next, add this category
    all_AFs['db'].extend([db] * len(af_i))
    all_AFs['genes'].extend([term] * len(af_i))
    all_AFs['AF'].extend([float(af) for af in af_i])
    
all_AFs = pd.DataFrame(all_AFs)

```

```python
def get_genes_dos_pval(d, dos, pval):
    ## returns list of genes from dataframe according to requested thresholds
    
    if dos == 'pos':
        return set(d[(d['dos'] > 0) & (d['mk.raw.p.value'] < pval)]['gene'])
    if dos == 'neg':
        return set(d[(d['dos'] < 0) & (d['mk.raw.p.value'] < pval)]['gene'])
```

```python
def make_position(dfi, col_chr, col_start):
    ## sorts by chromosome number and adds absolute postion column
    
    dfi = dfi.dropna()
    exclude = ['chrZ', 'chrW', 'chrNA', 'chrMT']
    dfi['COUNT'] = dfi[col_chr].str.replace('chr', '')
    
    dfi_num = dfi[ ~dfi[col_chr].isin(exclude)]
    dfi_Z = dfi[dfi[col_chr] == 'chrZ']
    dfi_W = dfi[dfi[col_chr] == 'chrW']
    
    dfi_num['COUNT'] = pd.to_numeric(dfi_num['COUNT'])
    dfi_num = dfi_num.sort_values(by=['COUNT', col_start])
    
    dfi_merged = pd.concat([dfi_num, dfi_Z, dfi_W], ignore_index=True)
    dfi_merged['POSITION'] = [x / 1000 for x in range(dfi_merged.shape[0])]
    return dfi_merged
```
